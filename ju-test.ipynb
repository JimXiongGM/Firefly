```python
from datasets import load_dataset

# 读取Parquet文件
file_path = './mlabonne/orpo-dpo-mix-40k/train-00000-of-00001.parquet'
dataset = load_dataset('parquet', data_files=file_path, split='train')
dataset = dataset.filter(
    lambda r: r["source"] != "toxic-dpo-v0.2"
)
```

```python
from transformers import AutoTokenizer

tokenizer = AutoTokenizer.from_pretrained("meta-llama/Meta-Llama-3.1-8B-Instruct")
```

```python
msg = [
        {
            "role": "system",
            "content": "You are an AI assistant."
        },
        {
            "role": "user",
            "content": "Hello, how are you?"
        },
        {
            "role": "assistant",
            "content": "I am doing well, thank you."
        }
    ]
print(tokenizer.apply_chat_template(msg, tokenize=False))
```
